Fusion des datasets

```python
train_df = dataframes['application_train.csv'].copy()
test_df = dataframes['application_test.csv'].copy()

def aggregate_and_merge_all(main_df, dict_dfs):
    config = {
        'bureau.csv': 'BUREAU',
        'previous_application.csv': 'PREV',
        'pos_cash_balance.csv': 'POS',
        'installments_payments.csv': 'INST',
        'credit_card_balance.csv': 'CREDIT'
    }
    
    for file_name, prefix in config.items():
        actual_key = next((k for k in dict_dfs.keys() if k.lower() == file_name.lower()), None)
        
        if actual_key:
            print(f"Agrégation et fusion de : {actual_key}...")
            source_df = dict_dfs[actual_key]
            
            # 1. Agrégation des numériques
            num_df = source_df.select_dtypes(include=['number']).copy()
            ids_to_drop = [c for c in num_df.columns if 'SK_ID' in c and c != 'SK_ID_CURR']
            num_agg = num_df.drop(columns=ids_to_drop).groupby('SK_ID_CURR').agg(['count', 'mean', 'max', 'min', 'sum']).reset_index()
            num_agg.columns = [f"{prefix}_{c[0]}_{c[1]}" if c[0] != 'SK_ID_CURR' else c[0] for c in num_agg.columns]
            
            # 2. Agrégation des catégorielles (AVEC SECURITE)
            cat_cols = source_df.select_dtypes(include=['object']).columns
            if len(cat_cols) > 0:
                cat_df = pd.get_dummies(source_df[['SK_ID_CURR'] + list(cat_cols)])
                cat_agg = cat_df.groupby('SK_ID_CURR').mean().reset_index()
                cat_agg.columns = [f"{prefix}_{c}" if c != 'SK_ID_CURR' else c for c in cat_agg.columns]
                final_agg = num_agg.merge(cat_agg, on='SK_ID_CURR', how='outer')
            else:
                final_agg = num_agg
                
            # 3. Fusion (Left Join)
            main_df = main_df.merge(final_agg, on='SK_ID_CURR', how='left')
            
    return main_df

# Exécution
train_df = aggregate_and_merge_all(train_df, dataframes)
test_df = aggregate_and_merge_all(test_df, dataframes)

print(f"\nDimensions finales du Train : {train_df.shape}")
print(f"Lignes (doit être 307511) : {train_df.shape[0]}")
```

## Structure du Jeu de Données Home Credit
### 1. Le Cœur du Projet
application_train.csv / application_test.csv :

C'est la table principale de l'analyse.
Contenu : Chaque ligne représente une demande de prêt spécifique. On y trouve les informations statiques du client au moment de la demande (âge, revenus, situation familiale, type d'emploi).
Cible : Le fichier train contient la colonne TARGET (0 : remboursé, 1 : difficulté de paiement).

### 2. Historique Externe (Le Bureau du Crédit)
bureau.csv :
Regroupe les crédits antérieurs contractés par le client auprès d'autres institutions financières.
Un client (SK_ID_CURR) peut avoir plusieurs lignes s'il a eu plusieurs crédits ailleurs.

bureau_balance.csv :
C'est le zoom mensuel des crédits listés dans le bureau.
Il permet de voir l'historique de l'état du crédit (fermé, actif, etc.) mois par mois.

### 3. Historique Interne (Prêts précédents chez Home Credit)
Ces fichiers détaillent les interactions passées du client avec l'organisme actuel.

previous_application.csv :
Historique de toutes les demandes de prêt passées du client chez Home Credit. On y voit si elles ont été approuvées, annulées ou refusées.

pos_cash_balance.csv :
Détail mensuel des anciens crédits à la consommation ou prêts en espèces.

credit_card_balance.csv :
Suivi mensuel des soldes et limites des cartes de crédit possédées par le client chez Home Credit.

installments_payments.csv :
Historique complet des paiements. C'est ici que l'on détecte si le client a l'habitude de payer avec du retard par rapport au calendrier prévu.

### Gestion des nan

```python
def missing_values_table(df):
    # Nombre total de valeurs manquantes
    mis_val = df.isnull().sum()
    
    # Pourcentage de valeurs manquantes
    mis_val_percent = 100 * df.isnull().sum() / len(df)
    
    # Création du tableau de synthèse
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis=1)
    
    # Renommer les colonnes
    mis_val_table_ren_columns = mis_val_table.rename(
    columns = {0 : 'Valeurs Manquantes', 1 : '% du Total'})
    
    # Trier par pourcentage descendant (en gardant seulement les colonnes avec des manques)
    mis_val_table_ren_columns = mis_val_table_ren_columns[
        mis_val_table_ren_columns.iloc[:,1] != 0].sort_values(
    '% du Total', ascending=False).round(1)
    
    # Affichage des infos
    print (f"Le DataFrame a {df.shape[1]} colonnes.\n"      
           f"Il y a {mis_val_table_ren_columns.shape[0]} colonnes avec des valeurs manquantes.")
    
    return mis_val_table_ren_columns

# On applique la fonction au fichier train
missing_values = missing_values_table(train_df)
missing_values.head(20)
```

### Suppression des colonnes avec trop de Nan

```python
threshold = 75

# On récupère le tableau des valeurs manquantes que tu as déjà calculé
missing_rows = missing_values[missing_values['% du Total'] > threshold]
cols_to_drop = list(missing_rows.index)

# Suppression dans les deux DataFrames
train_df = train_df.drop(columns = cols_to_drop)
test_df = test_df.drop(columns = cols_to_drop)

print(f"Suppression de {len(cols_to_drop)} colonnes ayant plus de {threshold}% de NaN.")
print(f"Colonnes supprimées (exemples) : {cols_to_drop[:5]}")
```

### Encodage

```python
from sklearn.preprocessing import LabelEncoder

# Définition des DataFrames
le = LabelEncoder()
le_count = 0

# Label Encoding
for col in train_df:
    if train_df[col].dtype == 'object':
        if len(list(train_df[col].unique())) <= 2:
            le.fit(train_df[col])
            train_df[col] = le.transform(train_df[col])
            test_df[col] = le.transform(test_df[col])
            le_count += 1

print('%d columns were label encoded.' % le_count)

# One-Hot Encoding
train_df = pd.get_dummies(train_df)
test_df = pd.get_dummies(test_df)

# Alignement
train_labels = train_df['TARGET']
train_df, test_df = train_df.align(test_df, join = 'inner', axis = 1)
train_df['TARGET'] = train_labels

print('Dimensions Train:', train_df.shape)
print('Dimensions Test:', test_df.shape)
```

### Fortes corrélations

```python
corr_matrix = train_df.drop(columns=['TARGET', 'SK_ID_CURR']).corr().abs()

# 2. On ne garde que le triangle supérieur pour éviter les doublons (A-B et B-A)
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# 3. Extraction de toutes les paires avec corrélation > 0.8
pd.set_option('display.max_rows', None) # Pour tout voir
pairs = upper.unstack()
strong_pairs = pairs[pairs > 0.8].sort_values(ascending=False).dropna().reset_index()
strong_pairs.columns = ['Feature_A', 'Feature_B', 'Correlation']

# 4. Affichage
print(f"Nombre de paires fortement colinéaires : {len(strong_pairs)}")
display(strong_pairs)

# Optionnel : réinitialiser l'affichage
pd.reset_option('display.max_rows')
```

### Suppression des fortes corrélations

```python
# 1. Calcul des corrélations avec la TARGET pour faire le bon choix
target_corrs = train_df.corr()['TARGET'].abs().sort_values(ascending=False)

# 2. Identifier les colonnes à supprimer parmi les paires
cols_to_drop = set()

for index, row in strong_pairs.iterrows():
    feat_a = row['Feature_A']
    feat_b = row['Feature_B']
    
    # Si aucune des deux n'est déjà marquée pour suppression
    if feat_a not in cols_to_drop and feat_b not in cols_to_drop:
        # On compare leur corrélation avec la cible
        corr_a = target_corrs.get(feat_a, 0)
        corr_b = target_corrs.get(feat_b, 0)
        
        # On supprime la moins corrélée à la cible
        if corr_a < corr_b:
            cols_to_drop.add(feat_a)
        else:
            cols_to_drop.add(feat_b)

# 3. Suppression effective
train_df = train_df.drop(columns=cols_to_drop)
test_df = test_df.drop(columns=cols_to_drop)

print(f"Nettoyage terminé : {len(cols_to_drop)} colonnes supprimées pour réduire la colinéarité.")
print(f"Dimensions finales du Train : {train_df.shape}")
```
